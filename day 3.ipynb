```python
#Exercise 9: Duplicate Class Filter
raw_classes = ["CSE", "ECE", "CSE", "IT", "ME", "ECE"]
unique_classes = set(raw_classes)
unique_classes.add("Civil")
print(unique_classes)
```

```python
#Exercise 10: Common Feature Selector
node_1 = {"temperature", "humidity", "pressure"}
node_2 = {"humidity", "pressure", "voltage", "lux"}
common_features = node_1.intersection(node_2)
unique_to_node_1 = node_1.difference(node_2)
print("Common features:", common_features)
print("Unique to node_1:", unique_to_node_1)
```

```python
#Exercise 11: Admission Eligibility Classifier
def check_eligibility(cgpa, entrance_score):
    if cgpa >= 7.5 and entrance_score >= 60:
        return "Eligible"
    else:
        return "Not Eligible"
result = check_eligibility(8.1, 65)
print(result)
```

```python
#Exercise 12: Predictor with Default Threshold
def predict_risk(probability, threshold=0.5):
    if probability >= threshold:
        return 1
    else:
        return 0
result1 = predict_risk(0.62)
result2 = predict_risk(0.62, 0.70)
print(result1)
print(result2)
```

```python
#Exercise 13: Variable Feature Aggregator *args
def calculate_mean(*scores):
    return sum(scores) / len(scores)
print(calculate_mean(80, 70, 90))
print(calculate_mean(80, 70, 90, 85, 75))
```
